# Importando libs

```python
import pandas as pd
import plotly.express as px
```

# Lendo o dataset

```python
df = pd.read_csv('../data/preprocessed/student_completion_data.csv')
```

# Pré-processando

```python
df_chars = df.copy()
```

```python
if df_chars["Completed"].dtype != "int64" and df_chars["Completed"].dtype != "float64":
    df_chars["Completed"] = df_chars["Completed"].map({
        "Yes": 1, "No": 0,
        "Sim": 1, "Não": 0,
        "Completed": 1, "Not Completed": 0,
        True: 1, False: 0
    })

df_chars["Completed_Label"] = df_chars["Completed"].map({1: "Concluiu", 0: "Não concluiu"})
```

# Gráficos

## Boxplot: tempo gasto vs conclusão

```python
fig_time_spent = px.box(
    df_chars,
    x="Completed_Label",
    y="Time_Spent_Hours",
    color="Completed_Label",
    title="Tempo total dedicado ao curso vs conclusão",
    labels={
        "Completed_Label": "Status do curso",
        "Time_Spent_Hours": "Horas totais dedicadas"
    }
)
fig_time_spent.show()
```

## Boxplot: dias desde último login

```python
fig_last_login = px.box(
    df_chars,
    x="Completed_Label",
    y="Days_Since_Last_Login",
    color="Completed_Label",
    title="Dias desde o último acesso vs conclusão",
    labels={
        "Completed_Label": "Status do curso",
        "Days_Since_Last_Login": "Dias desde o último login"
    }
)
fig_last_login.show()
```

## Histograma: taxa de conclusão de vídeos

```python
fig_video = px.histogram(
    df_chars,
    x="Video_Completion_Rate",
    color="Completed_Label",
    barmode="overlay",
    opacity=0.6,
    nbins=30,
    title="Distribuição da taxa de conclusão de vídeos",
    labels={
        "Video_Completion_Rate": "Taxa de vídeos concluídos (%)",
        "Completed_Label": "Status do curso"
    }
)
fig_video.show()
```

## Scatter: esforço vs desempenho

```python
fig_effort_perf = px.scatter(
    df_chars,
    x="Time_Spent_Hours",
    y="Quiz_Score_Avg",
    color="Completed_Label",
    title="Relação entre esforço e desempenho",
    labels={
        "Time_Spent_Hours": "Horas totais dedicadas",
        "Quiz_Score_Avg": "Média das notas em quizzes",
        "Completed_Label": "Status do curso"
    },
    opacity=0.7
)
fig_effort_perf.show()
```

## Boxplot: atividades perdidas

```python
fig_missed = px.box(
    df_chars,
    x="Completed_Label",
    y="Assignments_Missed",
    color="Completed_Label",
    title="Atividades perdidas vs conclusão",
    labels={
        "Completed_Label": "Status do curso",
        "Assignments_Missed": "Quantidade de atividades perdidas"
    }
)
fig_missed.show()
```

## Heatmap de correlação (variáveis numéricas)

```python
vars_corr = [
    "Login_Frequency",
    "Time_Spent_Hours",
    "Video_Completion_Rate",
    "Assignments_Submitted",
    "Assignments_Missed",
    "Quiz_Score_Avg",
    "Days_Since_Last_Login",
    "Completed"
]

corr = df_chars[vars_corr].corr()

fig_corr = px.imshow(
    corr,
    text_auto=".2f",
    aspect="auto",
    title="Mapa de correlação entre engajamento, desempenho e conclusão",
    color_continuous_scale="RdBu_r"
)
fig_corr.show()
```
